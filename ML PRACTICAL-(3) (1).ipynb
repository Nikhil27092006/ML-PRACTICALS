```python
X=df.iloc[:,:-1].values

y=df.iloc[:,-1].values
```

```python
def find_s(X,y):
    hypo= None
    print("Intial Hypothesis:",hypo)
    for i in range(len(X)):
        if y[i] == "acceptable":
            if hypo is None:
                hypo=X[i].copy()
            else:
                for j in range(len(hypo)):
                    if hypo[j] != X[i][j]:
                        hypo[j]="?"
        print(f"\nAfter Training Example {i+1}")
        print(hypo)
    return hypo
final_hypo=find_s(X,y)
print("\nFinal Hypothesis")
```

```python
# CANDIDATE-ELIMINATION ALGORITHM 
```

```python
import copy 
def candidate_elimination(concepts, target):
    specific =concepts[0].copy()
    general=[["?" for i in range(len(specific))]]
    print("\nIntial S:", specific)
    print("Intial G:",general)
    for i, h in enumerate(concepts):
        if target[i]=="acceptable":
            for x in range(len(specific)):
                if h[x] != specific[x]:
                    specific[x]="?"
        else:
            for x in range(len(specific)):
                if h[x] != specific[x]:
                    general.append(["?" if j != x else specific[x] for j in range(len(specific))])
        print("\nExample", i+1)
        print("S=",specific)
        print("G=",general)
    return specific, general
S,G= candidate_elimination(X,y)
print("\nFinal Specific Boundary")
print(S)
print("\nFinal General Boundary")
for g in G:
    print(g)
```
